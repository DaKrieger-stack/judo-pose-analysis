In this notebook, an RNN model will be trained using a **GRU (Gated Recurrent Unit)** architecture. GRUs are chosen instead of a standard RNN because they mitigate the vanishing gradient problem and capture longer temporal dependencies more effectively. Compared to **LSTM (Long Short-Term Memory)** networks, GRUs have a simpler structure with fewer parameters, which reduces computational cost and training time. This makes them suitable for a **Minimum Viable Product (MVP)** where faster experimentation and lower model complexity are beneficial while still maintaining stronger sequence modeling capabilities than a basic RNN.
